```python
import cv2 # OpenCV library
# 1 Load both images as greyscale (pixels 0–255)
img1 = cv2.imread("ao.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("ao2.jpg", cv2.IMREAD_GRAYSCALE)
# 2 Create the detector: keep the 500 best keypoints
orb = cv2.ORB_create(nfeatures=500)
# 3 Stages 0–2: pyramid, FAST, orientation, rBRIEF
kp1, des1 = orb.detectAndCompute(img1, None)
kp2, des2 = orb.detectAndCompute(img2, None)
# 4 Stage 3: compare bit strings by Hamming distance
bf = cv2.BFMatcher(cv2.NORM_HAMMING)
pairs = bf.knnMatch(des1, des2, k=2) # 2 nearest each
# 5 Ratio test: keep a match only if it clearly beats #2
good = [p[0] for p in pairs
if len(p) == 2 and p[0].distance < 0.75 * p[1].distance]
out = cv2.drawMatches(img1, kp1, img2, kp2, good, None)
print(len(kp1), des1.shape, len(good)) # e.g. 500 (500, 32
```

```python
import cv2
import matplotlib.pyplot as plt

img1_color = cv2.imread('/content/ao.jpg')
img2_color = cv2.imread('/content/ao2.jpg')

img1_rgb = cv2.cvtColor(img1_color, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2_color, cv2.COLOR_BGR2RGB)

img1_gray = cv2.cvtColor(img1_color, cv2.COLOR_BGR2GRAY)
img2_gray = cv2.cvtColor(img2_color, cv2.COLOR_BGR2GRAY)
orb = cv2.ORB_create(nfeatures=500)

kp1, des1 = orb.detectAndCompute(img1_gray, None)
kp2, des2 = orb.detectAndCompute(img2_gray, None)

bf = cv2.BFMatcher(cv2.NORM_HAMMING)
pairs = bf.knnMatch(des1, des2, k=2)

good = [p[0] for p in pairs if len(p) == 2 and p[0].distance < 0.75 * p[1].distance]

out_img = cv2.drawMatches(img1_rgb, kp1, img2_rgb, kp2, good, None, flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)

plt.figure(figsize=(15, 8))
plt.imshow(out_img)
plt.title(f'ORB Feature Matching (Matches: {len(good)})')
plt.axis('off')
plt.show()
```

Reflection & Discussion
1. Feature Effectiveness
Which features were the most useful?
Highly distinctive, high-contrast local structures—such as the sharp edges of the Ultraman's head, the bright boundary of the chest light, and limb joints—were the most useful.
Why did they work so well?
These areas contain sharp intensity gradients changing in multiple directions (strong eigenvalues in the image gradient matrix). This gives them high uniqueness and precise geometric localizability, resulting in highly distinct and reliable descriptors.
2. Ambiguous Features
Did any features cause ambiguity or false matches?
Yes. Symmetrical patterns on the Ultraman’s body (like matching left and right arm details) and uniform background colors can easily lead to similar ORB descriptors. Most of these ambiguities were filtered out by the strict ratio test threshold, leaving only 2 robust matches.
3. Robustness Challenge
Rotation: ORB remains highly effective because it computes an intensity centroid orientation for each keypoint, rendering the descriptors rotation-invariant.
Lighting changes: Extreme or non-uniform light sources (such as the bright glow background in the second image) distort local gradient patterns. This significantly degrades descriptor matching quality and reduces the number of successful matches.
